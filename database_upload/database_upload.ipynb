```python
# 1. Load credentials from .env
load_dotenv()

user = os.getenv('user')
password = os.getenv('password')
host = os.getenv('host')
port = os.getenv('port')
dbname = os.getenv('dbname')

# 2. Create the connection string for Supabase (PostgreSQL)
# Format: postgresql://username:password@host:port/database
connection_string = f"postgresql://{user}:{password}@{host}:{port}/{dbname}"
engine = create_engine(connection_string)

# 3. Define file paths (using relative paths based on your folder structure)
# Based on your image, we need to go up one level from 'database_upload' 
# then into 'data/original/processed'
processed_path = os.path.join(path, 'data', 'original')

files_to_upload = {
    'meal_info': 'meal_info.csv',
    'restaurant_train': 'restaurant_train.csv'
}

# 4. Loop through and upload
for table_name, file_name in files_to_upload.items():
    file_path = os.path.join(processed_path, file_name)
    
    print(f"Reading {file_name}...")
    df = pd.read_csv(file_path)
    
    print(f"Uploading to table '{table_name}' in Supabase...")
    # 'if_exists=replace' will create the table if it doesn't exist 
    # or overwrite it if it does.
    df.to_sql(table_name, engine, if_exists='replace', index=False)
    print(f"Successfully uploaded {table_name}!")

print("\nAll uploads complete.")
```
